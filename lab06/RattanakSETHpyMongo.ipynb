### Task 1: Create a Database and Insert Documents
Create a database called store and a collection called products. Insert at least three product
documents with fields for name, price, quantity, and category.

```python
from pymongo import MongoClient
# Connect to MongoDBto MongoDB
# client = MongoClient('mongodb://localhost:27017')
# db = client['store']
collection = db['products']
# Insert multiple product documents
products = [
{"name": "Laptop", "price": 999.99, "quantity": 10, "category":
"Electronics"},
{"name": "Coffee Mug", "price": 9.99, "quantity": 50, "category":
"Kitchen"},
{"name": "Headphones", "price": 49.99, "quantity": 25, "category":
"Electronics"}
]
collection.insert_many(products)
print("Products inserted successfully!")
```

```python
# Verify the insertion by checking the collection:
for product in collection.find():
    print(product)
```

### Task 2: Read Documents
Query the products collection to:
1. Retrieve all products in the "Electronics" category.
2. Find products with a price greater than $20.

```python
# Query for Electronics category
electronics = collection.find({"category": "Electronics"})
print("Electronics Products:")
for product in electronics:
    print(product)
# Query for products with price > $20
expensive_products = collection.find({"price": {"$gt": 20}})
print("\nProducts with price > $20:")
for product in expensive_products:
    print(product)

# Challenge: Modify the query to find products with a quantity less than 30 and in the"Electronics" category.
low_stock_electronics = collection.find({"quantity": {"$lt": 30}, "category": "Electronics"})
print("\nLow stock Electronics Products:")
for product in low_stock_electronics:
    print(product)
```

### Task 3: Update Documents
Update the quantity of "Laptop" to 15 and increase the price of all "Electronics" products by
10%.

```python
# Update quantity of Laptop
collection.update_one({"name": "Laptop"}, {"$set": {"quantity": 15}})
print("Laptop quantity updated!")
# Increase price of all Electronics products by 10%
collection.update_many({"category": "Electronics"}, {"$inc": {"price": 0.1 *
collection.find_one({"category": "Electronics"})["price"]}})
print("Electronics prices updated!")
# Verify updates
for product in collection.find():
    print(product)

# Challenge: Update the category of "Coffee Mug" to "Home Goods".
collection.update_one({"name": "Coffee Mug"}, {"$set": {"category": "Home Goods"}})
print("Coffee Mug category updated!")
# Verify the update
for product in collection.find({"name": "Coffee Mug"}):
    print(product)
```

```python
# Aggregation - Calculate the total value of inventory (price * quantity) for each category
# using MongoDB's aggregation pipeline. Hint: Use aggregate() with $group and $sum.
pipeline = [
    {
        "$group": {
            "_id": "$category",
            "total_value": {"$sum": {"$multiply": ["$price", "$quantity"]}}
        }
    }
]
result = collection.aggregate(pipeline)
print("\nTotal value of inventory by category:")
for category in result:
    print(category)

# Error Handling - Add try-except blocks to handle connection errors or invalid queries.
# try:
#     # Attempt to connect to MongoDB
#     client = MongoClient('mongodb://localhost:27017')
#     db = client['store']
#     collection = db['products']
#     print("Connected to MongoDB successfully!")
# except Exception as e:
#     print("Error connecting to MongoDB:", e)

# Complex Queries - Query for products with a price between $10 and $100 and sort them by price in descending order.
complex_query = collection.find({"price": {"$gte": 10, "$lte": 100}}).sort("price", -1)
print("\nProducts with price between $10 and $100 (sorted by price descending):")
for product in complex_query:
    print(product)
```

### Task 4: Delete Documents
Delete all products with a quantity less than 20.

```python
# Delete products with quantity < 20
result = collection.delete_many({"quantity": {"$lt": 20}})
print(f"Deleted {result.deleted_count} products.")
# Verify deletion
for product in collection.find():
    print(product)

# Challenge - Delete a specific product by name, e.g., "Headphones".
result = collection.delete_one({"name": "Headphones"})
print(f"Deleted {result.deleted_count} product(s) with name 'Headphones'.")
# Verify deletion
for product in collection.find():
    print(product)
# Close the connection
client.close()
```
