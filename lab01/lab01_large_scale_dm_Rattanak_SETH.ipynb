### Step 1: Initialize a SparkSession

```python
from pyspark.sql import SparkSession
# Create a SparkSession
spark = SparkSession.builder \
.appName("PySpark Lab Exercise") \
.getOrCreate()
# Print the SparkSession to confirm it’s working
print(spark)
```

### Step 2: Create a DataFrame
Create a small dataset and convert it into a PySpark DataFrame.

```python
# Sample data: list of tuples (name, department, salary)
data = [("Alice", "HR", 50000),
("Bob", "Engineering", 75000),
("Cathy", "Marketing", 60000)]
# Define column names
columns = ["Name", "Department", "Salary"]
# Create a DataFrame
df = spark.createDataFrame(data, columns)
# Display the DataFrame
df.show()
```

### Step 3: Perform Data Transformations
Explore basic PySpark operations by filtering and adding a column.

```python
# Filter employees with salary greater than 55000
high_earners = df.filter(df.Salary > 55000)
print("Employees with salary > 55000:")
high_earners.show()
# Add a column for a 10% bonus
df_with_bonus = df.withColumn("Bonus", df.Salary * 0.10)
print("DataFrame with bonus column:")
df_with_bonus.show()

# add it to df
df = df_with_bonus
# df.show()
```

```python
# 2. Modify the filter to show employees in the "Engineering" department only. (Hint: Use df.Department == "Engineering")
```

```python
# Filter employees in Engineering only
high_earners = df.filter(df.Department == "Engineering")
print("Employees in Engineering:")
high_earners.show()
```

### Step 4: Explore the DataFrame
Learn about the DataFrame’s structure and size.

```python
# Count the number of rows
row_count = df.count()
print(f"Total number of employees: {row_count}")
# Display the schema
print("DataFrame schema:")
df.printSchema()
```

### Challenge Questions

```python
# 1. Add a new column called Total_Compensation that sums Salary and Bonus. (Hint: Use df.withColumn and df.Salary + df.Bonus.)
df = df.withColumn("Total_Compensation", df.Salary + df.Bonus)
print("DataFrame with Total_Compensation column:")
df.show()
```

```python
# 2. Sort the DataFrame by Salary in descending order. (Hint: Use df.orderBy(df.Salary.desc()).)
df = df.orderBy(df.Salary.desc())
print("DataFrame sorted by Salary in descending order:")
df.show()
```

```python
# 3. Filter the DataFrame to show only employees whose names start with "A". (Hint: Use df.Name.startswith("A").)
df_a = df.filter(df.Name.startswith("A"))
print("Employees whose names start with 'A':")
df_a.show()
```
