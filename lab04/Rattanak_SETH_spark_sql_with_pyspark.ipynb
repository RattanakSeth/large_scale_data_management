```python
from pyspark.sql import SparkSession
```

```python
# Initialize Spark session
spark = SparkSession.builder \
.appName("SparkSQLLab") \
.getOrCreate()
```

### Load Data into Spark
Add the following code to your script to load the JSON files and
create temporary views for SQL queries.

```python
# Load JSON data into DataFrames
employees_df = spark.read.option("multiline","true") \
      .json("employees.json")
departments_df = spark.read.option("multiline", "true") \
    .json("departments.json")

# Create temporary views for SQL queries
employees_df.createOrReplaceTempView("employees")
departments_df.createOrReplaceTempView("departments")
```

### Lab Tasks
Complete the following tasks using Spark SQL. Write SQL queries and execute them using
spark.sql(). For each task, display the results using .show().

### Task 1: Basic Selection
Write a query to select all columns from the employees table.

```python
# employees_df.printSchema()
result = spark.sql('SELECT * FROM employees')
result.show()
```

### Task 2: Filtering
Write a query to find employees with a salary greater than 65,000.

```python
result = spark.sql('SELECT name, salary FROM employees WHERE salary > 65000')
result.show()
```

### Task 3: Aggregation
Write a query to calculate the average salary per department (dept_id).

```python
result = spark.sql("""
    SELECT dept_id, AVG(salary) as avg_salary 
    FROM employees 
    GROUP BY dept_id
""")
result.show()
```

### Task 4: Join
Write a query to join the employees and departments tables and display the employee name,
department name, and salary.

```python
result4 = spark.sql("""
    SELECT e.name, d.dept_name , e.salary
    FROM employees e 
    JOIN departments d ON e.dept_id = d.dept_id
""")
result4.show()
```

### Task 5: Complex Query
Write a query to find the department(s) with the highest average salary. Use a subquery or
common table expression (CTE).

```python
result = spark.sql("""
    WITH dept_avg_salary AS (
        SELECT dept_id, AVG(salary) AS avg_salary
        FROM employees
        GROUP BY dept_id
    )
    SELECT dept_id, avg_salary
    FROM dept_avg_salary
    WHERE avg_salary = (SELECT MAX(avg_salary) FROM dept_avg_salary)
""")
result.show()
```
